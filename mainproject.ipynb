```python
import oracledb
import pandas as pd
from sqlalchemy import create_engine, types
from sqlalchemy.dialects import oracle

# Step 1 - Init Oracle client
oracledb.init_oracle_client(lib_dir=r"C:\oracle\instantclient_23_0")

# Step 2 - Connect
engine = create_engine(
    "oracle+oracledb://hr:hr@localhost:1521/?service_name=xe"
)

# Step 3 - Map DataFrame dtypes to Oracle-compatible types
def get_oracle_dtype_map(df):
    dtype_map = {}
    for col, dtype in df.dtypes.items():
        if dtype == 'float64' or dtype == 'float32':
            dtype_map[col] = types.Numeric(precision=18, scale=4)  # Oracle-safe
        elif dtype == 'int64' or dtype == 'int32':
            dtype_map[col] = types.Integer()
        elif dtype == 'object':
            dtype_map[col] = types.VARCHAR(255)
        elif dtype == 'datetime64[ns]':
            dtype_map[col] = types.DateTime()
    return dtype_map

# Step 4 - Load DataFrame into Oracle
df.to_sql(
    name='cust_shopping',
    con=engine,
    if_exists='replace',
    index=False,
    dtype=get_oracle_dtype_map(df)   # Explicit type mapping
)

print(" Table loaded into Oracle successfully!")

# Step 5 - Verify
df_check = pd.read_sql("SELECT * FROM cust_shopping", con=engine)
print(df_check.head())
```
